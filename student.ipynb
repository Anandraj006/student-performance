**Visualization**


```python
plt.figure(figsize=(9, 3))

df.isnull().sum().plot(kind='bar')

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=90)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['previous_exam_score'], bins=40)

plt.title("Previous Exam Score Distribution")
plt.xlabel("Previous Exam Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['previous_gpa'].dropna(), bins=10)

plt.title("Previous GPA Distribution")
plt.xlabel("Previous GPA")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['attendance_percentage'].dropna(), bins=10)

plt.title("Attendance Percentage Distribution")
plt.xlabel("Attendance Percentage")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['exam_score'].dropna(), bins=10)

plt.title("Exam Score Distribution")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['study_hours_per_day'],
    df['exam_score']
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours Per Day")
plt.ylabel("Exam Score")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['previous_gpa'],
    df['exam_score']
)

plt.title("Previous GPA vs Exam Score")
plt.xlabel("Previous GPA")
plt.ylabel("Exam Score")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['sleep_hours'],
    df['exam_score']
)

plt.title("Sleep Hours vs Exam Score")
plt.xlabel("Sleep Hours")
plt.ylabel("Exam Score")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['time_management_score'],
    df['exam_score']
)

plt.title("Time Management vs Exam Score")
plt.xlabel("Time Management Score")
plt.ylabel("Exam Score")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    df['questions_attempted'],
    df['questions_correct']
)

plt.title("Questions Attempted vs Questions Correct")
plt.xlabel("Questions Attempted")
plt.ylabel("Questions Correct")

plt.show()
```

```python
performance_count = df['performance_level'].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    performance_count.index,
    performance_count.values
)

plt.title("Student Performance Level")
plt.xlabel("Performance Level")
plt.ylabel("Number of Students")

plt.show()
```

```python
pass_count = df['pass_status'].value_counts()

plt.figure(figsize=(7, 5))

plt.bar(
    pass_count.index,
    pass_count.values
)

plt.title("Student Pass Status")
plt.xlabel("Pass Status")
plt.ylabel("Number of Students")

plt.show()
```

```python
gender_count = df['gender'].value_counts()

plt.figure(figsize=(7, 5))

plt.bar(
    gender_count.index,
    gender_count.values
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

correlation = numeric_df.corr()

plt.figure(figsize=(12, 8))

plt.imshow(correlation)

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Matrix")

plt.show()
```
